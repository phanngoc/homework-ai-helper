```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("lequidon/vi-math-10-grade")

print("Path to dataset files:", path)
```

```python
import os

# List all files in the dataset directory
files = os.listdir(path)
print("Files in dataset:")
for file in files:
    print(f"- {file}")
```

```python
import re
from collections import Counter

# Pattern to extract year from filename
year_pattern = r'nam-(\d{4})-(\d{4})'

# Initialize counters
year_counts = Counter()
province_counts = Counter()

# Analyze filenames
for file in files:
    if file.endswith('.pdf'):
        # Extract years
        year_match = re.search(year_pattern, file)
        if year_match:
            school_year = f"{year_match.group(1)}-{year_match.group(2)}"
            year_counts[school_year] += 1
            
        # Extract province (assuming last word before .pdf is province)
        province = file.split('-')[-1].replace('.pdf', '').title()
        if province.isalpha():  # Only count if it's a valid province name
            province_counts[province] += 1

print(f"Number of PDF files: {len([f for f in files if f.endswith('.pdf')])}")
print("\nSchool year distribution:")
for year, count in sorted(year_counts.most_common()):
    print(f"{year}: {count} files")

print("\nTop 10 provinces by number of files:")
for province, count in province_counts.most_common(10):
    print(f"{province}: {count} files")
```

```python
import PyPDF2

if len(files) > 0:
    sample_file = os.path.join(path, files[0])
    with open(sample_file, 'rb') as f:  # Open in binary mode for PDFs
        pdf_reader = PyPDF2.PdfReader(f)
        # Get text from first page
        if len(pdf_reader.pages) > 0:
            text = pdf_reader.pages[0].extract_text()
            print(f"\nSample content from {files[0]}:")
            print(text[:500])  # Print first 500 characters
```
